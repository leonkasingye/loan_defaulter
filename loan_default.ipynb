# Stage 2 — Data Understanding

 Dataset Overview
- **Rows:** 255,347
- **Columns:** 18 (10 numeric, 7 categorical, 1 identifier)
- **Target:** `Default` (binary: 0 = no default, 1 = default)

Target Distribution
| Class | Count | Percent |
|-------|-------|---------|
| 0 (No Default) | 225,694 | 88.4% |
| 1 (Default) | 29,653 | 11.6% |
- **Imbalance ratio:** ≈ 7.6 : 1
- **Consequence:** Accuracy is a misleading metric. We will use **recall** (catching defaulters) as the primary metric, with precision and AUC as secondary.


- **None** across all 18 columns.

 Data Types
- **10 numeric** — Age, Income, LoanAmount, CreditScore, MonthsEmployed, NumCreditLines, InterestRate, LoanTerm, DTIRatio, Default
- **7 categorical (str)** — Education, EmploymentType, MaritalStatus, HasMortgage, HasDependents, LoanPurpose, HasCoSigner
- **1 identifier (str)** — LoanID

 Numeric Summary
- All numeric ranges are within plausible real-world bounds.
- Mean ≈ median for Age, Income, CreditScore → roughly symmetric distributions.
- Suspiciously round maximums: Income ≤ 149,999; LoanAmount ≤ 249,999; CreditScore ≤ 849. Suggests the data is capped / pre-processed.
- `MonthsEmployed` has a minimum of 0 → worth investigating in Stage 3.

Categorical Summary
All 7 categorical columns are **clean** — no typos, no casing issues, no placeholder values.

| Column | Categories | Distribution |
|--------|-----------|--------------|
| Education | 4 (PhD, Master's, Bachelor's, High School) | ~uniform |
| EmploymentType | 4 (Full-time, Part-time, Self-employed, Unemployed) | ~uniform |
| MaritalStatus | 3 (Single, Married, Divorced) | ~uniform |
| HasMortgage | 2 (Yes, No) | 50/50 |
| HasDependents | 2 (Yes, No) | 50/50 |
| LoanPurpose | 5 (Auto, Business, Education, Home, Other) | ~uniform |
| HasCoSigner | 2 (Yes, No) | 50/50 |

Duplicates
- **None** (checked both with and without LoanID).

Dataset Nature — Important Caveat
The dataset appears **synthetic or heavily pre-processed**:
1. Categorical columns have near-perfect uniform distributions
2. Numeric columns have suspiciously round caps
3. `EmploymentType = "Unemployed"` appears in ~25% of rows (unrealistic for real lending)
4. Zero missing values, zero duplicates, zero typos

**Implication:** Model metrics may be inflated relative to real-world performance. The **methodology** we build is transferable; the **specific numbers** are not.

Hypotheses for Stage 3
- `LoanID` must be dropped — pure identifier
- Binary `Yes`/`No` columns → convert to 0/1
- Multi-category columns → one-hot encode (low cardinality, easy)
- Investigate `MonthsEmployed = 0` — placeholder or genuine?
- Possibly bin `Age` and `Income` for robustness

 Ready for Stage 3
Every cleaning decision in Stage 3 will be justified by the findings above.
