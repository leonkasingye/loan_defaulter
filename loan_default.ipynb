Step 3.3                   Investigate MonthsEmployed = 0
In Stage 2.5 we saw MonthsEmployed min = 0. Let's see how many rows have this value and whether they correlate with anything suspicious.



```python
# How many rows have MonthsEmployed = 0?
zero_emp = (df_clean["MonthsEmployed"] == 0).sum()
print(f"Rows with MonthsEmployed = 0: {zero_emp} ({zero_emp/len(df_clean)*100:.2f}%)")

# Look at a few of them
df_clean[df_clean["MonthsEmployed"] == 0].head()
```

The Follow-Up Question
Let's see whether MonthsEmployed = 0 correlates with EmploymentType. If most of them are "Unemployed" → it's a legitimate value. If they're spread across all employment types → it's a placeholder.

```python
# How do MonthsEmployed=0 rows distribute across EmploymentType?
zero_emp_rows = df_clean[df_clean["MonthsEmployed"] == 0]

print("Distribution of EmploymentType among MonthsEmployed=0 rows:")
print(zero_emp_rows["EmploymentType"].value_counts())
print()

print("Overall EmploymentType distribution for comparison:")
print(df_clean["EmploymentType"].value_counts(normalize=True).round(4) * 100)
```

Outlier check

```python
# Quick outlier scan — count how many values fall outside 1.5×IQR for each numeric col
numeric_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns.drop("Default")

for col in numeric_cols:
    q1 = df_clean[col].quantile(0.25)
    q3 = df_clean[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()
    pct = outliers / len(df_clean) * 100
    print(f"{col:20s} outliers: {outliers:6d} ({pct:5.2f}%)")
```

```python
zero_emp = (df_clean["MonthsEmployed"] == 0).sum()
print(f"Rows with MonthsEmployed = 0: {zero_emp} ({zero_emp/len(df_clean)*100:.2f}%)")

# Look at a few examples
df_clean[df_clean["MonthsEmployed"] == 0].head()
```

Logical Consistency Check


```python
# Define expected valid ranges for key columns
checks = {
    "Age": (18, 100),
    "CreditScore": (300, 850),
    "DTIRatio": (0.0, 1.0),
    "InterestRate": (0, 50),
    "LoanTerm": (1, 60),
    "MonthsEmployed": (0, 600),
}

for col, (low, high) in checks.items():
    invalid = ((df_clean[col] < low) | (df_clean[col] > high)).sum()
    print(f"{col:20s} outside [{low}, {high}]: {invalid}")
```

# Stage 4 — Feature Engineering


Reload the Cleaned Data
We start fresh from the frozen CSV. This ensures our session state is clean.

```python
import pandas as pd
import numpy as np

# Load the cleaned dataset from disk
df = pd.read_csv("data/loan_default_clean.csv")

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nDtypes summary:")
print(df.dtypes.value_counts())
```

Encoding Categoricals


```python
# ---- 1. Binary columns: Yes/No → 1/0 ----
binary_cols = ["HasMortgage", "HasDependents", "HasCoSigner"]
for col in binary_cols:
    df[col] = df[col].map({"Yes": 1, "No": 0})

# ---- 2. Ordinal column: Education ----
education_order = {
    "High School": 0,
    "Bachelor's": 1,
    "Master's": 2,
    "PhD": 3,
}
df["Education"] = df["Education"].map(education_order)

# ---- 3. Nominal columns: One-hot encode ----
nominal_cols = ["EmploymentType", "MaritalStatus", "LoanPurpose"]
df = pd.get_dummies(df, columns=nominal_cols, drop_first=True, dtype=int)

# ---- Verify ----
print("Shape after encoding:", df.shape)
print("\nRemaining non-numeric columns:",
      df.select_dtypes(include=["object", "str"]).columns.tolist())
print("\nAll columns now:")
for col in df.columns:
    print(f"  - {col}")
```
